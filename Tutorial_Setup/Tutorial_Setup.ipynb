# Tutorial Setup

Before our tutorial on uproot later, we're going to download a file. We'll run this now so that it's there for us once we're done.

```python
import uproot
import XRootD
import os

# Create XRootD client
import XRootD.client as xrdclient
eic_server = 'root://dtn-eic.jlab.org:1094/'
fs = xrdclient.FileSystem(eic_server)

fpath = '/volatile/eic/EPIC/RECO/26.02.0/epic_craterlake/DIS/NC/18x275/minQ2=10/'
fname = eic_server+fpath+'pythia8NCDIS_18x275_minQ2=10_beamEffects_xAngle=-0.025_hiDiv_1.0000.eicrecon.edm4eic.root'
outfname = "/Users/hszumila/Desktop/Day_1_Tutorial_Input.root" # Change this if you want
```

```python
status, files = fs.dirlist(fpath)
if status.ok: # Check we can see our files and connect OK
    if os.path.isfile(outfname): # Check if we've already downloaded our file
        print(outfname, "- already exists, no need to download it again!")
    else: # If we haven't, download it!
        print("Downloading file - ", fname)
        fs.copy(fname, outfname, False)
        print("File saved as -",outfname, "- All done!")
else:
  print(f"Error: {status.message}")
```
